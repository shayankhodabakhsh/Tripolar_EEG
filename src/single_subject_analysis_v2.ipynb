# 🧠 Tripolar EEG — Single Subject Analysis (Stable Figure Names)

This notebook is a **clean, reproducible** single-subject runner.

- Figures save into `output/<subject>/single_subject/` with **stable filenames** (re-running cells overwrites the same files).
- Uses the built-in plotting functions in `eeg_analysis.py` so filenames are meaningful (not `fig_###.png`).


```python
import sys
from pathlib import Path

# Make imports work whether you run from repo root or src/
_cwd = Path.cwd().resolve()
if (_cwd / 'data').exists():
    repo_root = _cwd
elif (_cwd.parent / 'data').exists():
    repo_root = _cwd.parent
else:
    repo_root = _cwd

sys.path.insert(0, str(repo_root / 'src'))

from eeg_analysis import *

%matplotlib inline
plt.rcParams['figure.dpi'] = 120

DATA_DIR = str((repo_root / 'data').resolve())
OUT_DIR = (repo_root / 'output')

print(f"DATA_DIR = {DATA_DIR}")
print(f"OUT_DIR  = {OUT_DIR}")
```

```python
# Discover subjects
subjects = discover_subjects(DATA_DIR)
print(f"Found {len(subjects)} subjects:\n")
for s in subjects:
    avg_status = "✓" if s['avg'] else "✗"
    vmrk_status = "✓" if s['vmrk'] else "✗"
    print(f"  {s['name']:<15s}  files: {s['basename']}")
    print(f"  {'':15s}  markers: {vmrk_status}   avg: {avg_status}")
    print()
```

```python
# Select a subject
SUBJECT = "CP1"  # <-- change me

print(f"Loading: {SUBJECT}")
subject = load_subject(DATA_DIR, SUBJECT)
results = analyze_subject(subject)

save_dir = OUT_DIR / subject['name'] / 'single_subject'
save_dir.mkdir(parents=True, exist_ok=True)

print(f"Saving figures to: {save_dir}")
print("Analysis complete ✓")
```

```python
# Summary dashboard (stable filename: <subject>_summary.png)
plot_subject_summary(subject, results, save_dir=str(save_dir), show=True)
```

```python
# Full figure set (stable filenames like: <subject>_psd.png, <subject>_spectrogram_ch1.png, ...)
# Re-running this cell overwrites the same files.
plot_subject_full(subject, results, save_dir=str(save_dir), show=True)
```

## Why your old notebook created `unknown_subject/` and changing filenames

In `single_subject_analysis.ipynb` the autosave hook:
- Saves to `Path('output')` **relative to the current working directory**. If you run the notebook from `src/`, you’ll get `src/output/...`.
- Uses `unknown_subject` until you explicitly call `set_plot_subject(...)`.
- Uses an incrementing counter (`fig_001`, `fig_002`, …), so re-running a cell makes new filenames.

This v2 notebook avoids all of that by using `plot_subject_summary(..., save_dir=...)` and `plot_subject_full(..., save_dir=...)`, which save **meaningful, stable filenames**.
